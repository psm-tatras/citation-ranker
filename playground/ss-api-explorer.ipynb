```python
API_KEY = "kMz68jcdLo7VWB5zpKGrv5D4rJu1lljE58tBB8z5"

def get_paper_details(title: str) -> dict | None:
    """Search for a paper by title and return its details."""
    
    base_url = "https://api.semanticscholar.org/graph/v1/paper/search"
    
    headers = {
        "x-api-key": API_KEY
    }
    
    # Fields to retrieve
    fields = ",".join([
        "title",
        "abstract",
        "year",
        "authors",
        "citationCount",
    ])
    
    params = {
        "query": title,
        "fields": fields,
        "limit": 1  # Get the top match
    }
    
    response = requests.get(base_url, headers=headers, params=params)
    response.raise_for_status()
    
    data = response.json()
    papers = data.get("data", [])
    
    if not papers:
        print("No papers found.")
        return None
    
    return papers[0]
```

```python
import requests
import os
from pathlib import Path

def download_dataset_files(data, download_dir="./data", show_progress=True):
    """Download all files from data['files'] (presigned S3 URLs)."""
    files = data["files"]
    Path(download_dir).mkdir(parents=True, exist_ok=True)

    for i, url in enumerate(files, 1):
        filename = url.split("?")[0].split("/")[-1]  # strip query params
        dest = os.path.join(download_dir, filename)

        if os.path.exists(dest):
            print(f"[{i}/{len(files)}] Skipping {filename} (already exists)")
            continue

        print(f"[{i}/{len(files)}] Downloading {filename}...")
        with requests.get(url, stream=True) as r:
            r.raise_for_status()
            with open(dest, "wb") as f:
                for chunk in r.iter_content(chunk_size=1024 * 1024):  # 1MB chunks
                    f.write(chunk)
        print(f"  -> Saved to {dest}")

# Usage (continuing from your last cell):
data = response.json()
download_dataset_files(data, download_dir="/Work/Experiments/IISER/s2orc-direct")
```
